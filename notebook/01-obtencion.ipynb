## Base de datos permiso de circulación 2022

```python
import pyodbc

# Lista los drivers instalados en tu sistema
for driver in pyodbc.drivers():
    if "Access" in driver:
        print(driver)
```

```python
import pyodbc

ruta_bd = r".\base-de-datos-permisos-de-circulación 2022.accdb"
conn_str = r"DRIVER={Microsoft Access Driver (*.mdb, *.accdb)};DBQ=" + ruta_bd

conexion = pyodbc.connect(conn_str)
cursor = conexion.cursor()

# Obtener los nombres de todas las tablas
tablas = [row.table_name for row in cursor.tables(tableType="TABLE")]
print("Tablas encontradas en la base de datos:")
for t in tablas:
    print(f"- {t}")

conexion.close()
```

```python
import pandas as pd
import pyodbc

ruta_bd = r".\base-de-datos-permisos-de-circulación 2022.accdb"

conn_str = (
    r"DRIVER={Microsoft Access Driver (*.mdb, *.accdb)};"
    f"DBQ={ruta_bd};"
)

conexion = pyodbc.connect(conn_str)
nombre_tabla_real = "TBL_VehículosConMotor"

query = f"SELECT * FROM [{nombre_tabla_real}]"
df2022 = pd.read_sql(query, conexion)

conexion.close()
print("los primeros 5 registros")
df2022.head()
```
