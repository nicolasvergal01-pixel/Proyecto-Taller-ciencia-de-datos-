```python
import pyodbc

ruta_bd = r"C:\Users\nvergal\Documents\emisiones gases 2\base-de-datos-permisos-de-circulación 2022.accdb"
conn_str = r"DRIVER={Microsoft Access Driver (*.mdb, *.accdb)};DBQ=" + ruta_bd

conexion = pyodbc.connect(conn_str)
cursor = conexion.cursor()

# Obtener los nombres de todas las tablas
tablas = [row.table_name for row in cursor.tables(tableType="TABLE")]
print("Tablas encontradas en la base de datos:")
for t in tablas:
    print(f"- {t}")

conexion.close()
```
